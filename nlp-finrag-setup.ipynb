### Step 1: Install Miniconda to manage Python environments safely
We will download and install Miniconda to establish an isolated Python 3.11 environment.

````python
%%bash
MINICONDA_INSTALLER_SCRIPT=Miniconda3-latest-Linux-x86_64.sh
MINICONDA_PREFIX=/usr/local
wget -q https://repo.anaconda.com/miniconda/$MINICONDA_INSTALLER_SCRIPT
chmod +x $MINICONDA_INSTALLER_SCRIPT
./$MINICONDA_INSTALLER_SCRIPT -b -f -p $MINICONDA_PREFIX
rm $MINICONDA_INSTALLER_SCRIPT
````

### Step 2: Create a Python 3.11 Environment
We'll update conda, configure channels, and create a python environment named `finrag` specifically targeting Python 3.11.

````python
%%bash
# Accept terms of service non-interactively
conda tos accept --override-channels --channel https://repo.anaconda.com/pkgs/main
conda tos accept --override-channels --channel https://repo.anaconda.com/pkgs/r

# Configure channels and create the python 3.11 environment
conda config --add channels defaults
conda config --add channels conda-forge
conda create -y --name finrag python=3.11
````

### Step 3: Install Poetry in the Python 3.11 environment
We will activate our environment and install poetry inside it. Then, we can run poetry-related installations.

````python
%%bash
# Install poetry directly inside the Python 3.11 environment
conda run -n finrag pip install poetry

# Verify that poetry is installed and pointing to Python 3.11
conda run -n finrag python -V
conda run -n finrag poetry --version
````

### Step 4: Install your project dependencies using Poetry
Now we can trigger `poetry install` inside our Python 3.11 virtual environment.

````python
%%bash
# Tell poetry to use the Python interpreter of the finrag environment
conda run -n finrag poetry env use $(conda run -n finrag which python)

# Install the project dependencies
conda run -n finrag poetry install
````
