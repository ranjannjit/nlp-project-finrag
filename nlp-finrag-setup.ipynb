### Step 1: Install Miniconda to manage Python environments safely
We will download and install Miniconda to establish an isolated Python 3.11 environment.

````python
%%bash
MINICONDA_INSTALLER_SCRIPT=Miniconda3-latest-Linux-x86_64.sh
MINICONDA_PREFIX=/usr/local
wget -q https://repo.anaconda.com/miniconda/$MINICONDA_INSTALLER_SCRIPT
chmod +x $MINICONDA_INSTALLER_SCRIPT
./$MINICONDA_INSTALLER_SCRIPT -b -f -p $MINICONDA_PREFIX
rm $MINICONDA_INSTALLER_SCRIPT
````

### Step 2: Create a Python 3.11 Environment
We'll update conda, configure channels, and create a python environment named `finrag` specifically targeting Python 3.11.

````python
%%bash
# Accept terms of service non-interactively
conda tos accept --override-channels --channel https://repo.anaconda.com/pkgs/main
conda tos accept --override-channels --channel https://repo.anaconda.com/pkgs/r

# Configure channels and create the python 3.11 environment
conda config --add channels defaults
conda config --add channels conda-forge
conda create -y --name finrag python=3.11
````

### Step 3: Install Poetry in the Python 3.11 environment
We will activate our environment and install poetry inside it. Then, we can run poetry-related installations.

````python
%%bash
# Install poetry directly inside the Python 3.11 environment
conda run -n finrag pip install poetry

# Verify that poetry is installed and pointing to Python 3.11
conda run -n finrag python -V
conda run -n finrag poetry --version
````

### Step 4: Install your project dependencies using Poetry
Now we can trigger `poetry install` inside our Python 3.11 virtual environment.

````python
%%bash
# Tell poetry to use the Python interpreter of the finrag environment
conda run -n finrag poetry env use $(conda run -n finrag which python)

# Install the project dependencies
conda run -n finrag poetry install
````

````python
import json

def clean_and_verify_json(filepath):
    with open(filepath, 'r', encoding='utf-8') as f:
        raw_data = f.read().strip()

    # Try to parse the file normally first
    try:
        data = json.loads(raw_data)
        print(f"Successfully parsed! JSON is already valid with {len(data)} items.")
        return
    except Exception as e:
        print(f"Standard parsing failed: {e}. Attempting recovery...")

    # If standard parsing fails with "Extra data", it means there are multiple json structures or arrays
    # We can use a raw JSON decoder to pull valid objects or arrays from the file sequentially.
    decoder = json.JSONDecoder()
    pos = 0
    all_items = []

    while pos < len(raw_data):
        # Skip whitespaces, commas, and list markers
        raw_data_sub = raw_data[pos:].lstrip(" \t\r\n,[]")
        if not raw_data_sub:
            break
        # Update position based on skipped characters
        pos = len(raw_data) - len(raw_data_sub)
        try:
            obj, idx = decoder.raw_decode(raw_data[pos:])
            if isinstance(obj, list):
                all_items.extend(obj)
            elif isinstance(obj, dict):
                all_items.append(obj)
            pos += idx
        except Exception as err:
            print(f"Stopping extraction due to error at position {pos}: {err}")
            break

    print(f"Extracted {len(all_items)} valid dictionary items from raw data.")

    # Write the clean unified list of objects back to the file
    with open(filepath, 'w', encoding='utf-8') as f:
        json.dump(all_items, f, indent=4)
    print("Unified JSON written successfully to source.")

clean_and_verify_json('/content/nlp-project-finrag/data/train.json')
````

````python
import json
import re

def repair_truncated_json(filepath):
    with open(filepath, 'r', encoding='utf-8') as f:
        content = f.read().strip()

    # Locate the last complete JSON block. Since it is a list of dicts, we find the last fully closed dictionary '}'
    # We want to find the last occurrence of '}' followed by spacing/newlines, a comma, and then an incomplete structure.
    # Let's clean up trailing commas and find the last matching closed element block.
    matches = list(re.finditer(r'\}\s*,?\s*(?:\{\s*"[^"]+"|\s*$)', content))
    if not matches:
        print("Could not find a clean block boundary to truncate and repair.")
        return

    # We truncate at the last complete dict match
    last_match = matches[-1]
    end_idx = last_match.start() + 1
    repaired_content = content[:end_idx]

    # Add the closing bracket for the main list
    repaired_content = repaired_content.rstrip().rstrip(',') + "\n]"

    # Let's verify if the repaired version loads correctly
    try:
        data = json.loads(repaired_content)
        print(f"Successfully repaired JSON! Loaded {len(data)} items.")
        with open(filepath, 'w', encoding='utf-8') as f:
            f.write(repaired_content)
    except Exception as e:
        print("Repair attempt failed to produce valid JSON:", str(e))

repair_truncated_json('/content/nlp-project-finrag/data/train.json')
````
